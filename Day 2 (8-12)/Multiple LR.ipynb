```python
#residual erro, mean square error , sum of sqaures , mean error , rms error
import pandas as pd
df = pd.read_csv("student_multi_scores.csv")
```

```python
df.columns
```

```python
x=df[['Hours', 'IQ', 'Sleep']]
y=df['Scores']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.score(x_train,y_train)
```

```python
print(y_pred)
print(y_test)
```

```python
import matplotlib.pyplot as plt
predictions = model.predict(x)
plt.scatter(y,predictions,color='blue')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', linewidth=4)
plt.xlabel("Actual Scores")
plt.ylabel("Predicted Scores")
plt.title("Actual vs Predicted (Dots on Line = Good)")
plt.show()
```
