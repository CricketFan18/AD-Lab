```python
cart=[]
print("MENU\n1-> View shop\n2->Add to cart\n3->remove item\n4->Go to Cart")
shop=["skirt","top","jeans","jacket","pants","shirt","tshirt"]
while True:
    n=int(input("enter choice:"))
    match n:
        case 1:
            print(shop)
        case 2:
             item=input("enter product name:")
             cart.append(item)
        case 3:
            item=input("enter product name:")
            cart.remove(item)
        case 4:
            print(cart)
    m=input("Do you wish to comtinue shopping?y/n")
    if m=='n':
        break
```
